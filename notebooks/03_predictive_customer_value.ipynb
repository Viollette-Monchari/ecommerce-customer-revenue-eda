```python
#load pandas/ numpy and commit

import pandas as pd
import numpy as np
```

```python
#show the cleaned customer summary as we are trying to answer the question: who are our most valued customers

customer_summary = pd.read_csv("../data/cleaned/customer_summary.csv")
customer_summary.head()
```

```python
#create the target variable (is_high_value_customer) - here just checking for column name

revenue_col = "TotalRevenue"
customer_summary[revenue_col].describe()
```

```python
#is_high_value_customer (top 25%)

threshold = customer_summary[revenue_col].quantile(0.75)

customer_summary["is_high_value_customer"] = (
    customer_summary[revenue_col] >= threshold
).astype(int)

customer_summary[[revenue_col, "is_high_value_customer"]].head()
```

```python
#choose features using the real column names and then commit

X = customer_summary[["NumInvoices"]]
y = customer_summary["is_high_value_customer"]

X.shape, y.value_counts()
```

```python
# 1) Create a rule threshold for NumInvoices
invoice_threshold = customer_summary["NumInvoices"].quantile(0.75)

# 2) Baseline prediction using only NumInvoices
customer_summary["pred_high_value_baseline"] = (
    customer_summary["NumInvoices"] >= invoice_threshold
).astype(int)

# 3) Quick check: how many predicted as high value?
customer_summary["pred_high_value_baseline"].value_counts()
```

```python
#evaluate baseline accuracy

# Actual labels
y_true = customer_summary["is_high_value_customer"]

# Baseline predicted labels
y_pred = customer_summary["pred_high_value_baseline"]

# Accuracy = how often baseline matches actual
accuracy = (y_true == y_pred).mean()
accuracy
```

```python
#Confusion Matrix (More informative than accuracy) and then commit


pd.crosstab(
    y_true,
    y_pred,
    rownames=["Actual"],
    colnames=["Predicted"]
)
```
